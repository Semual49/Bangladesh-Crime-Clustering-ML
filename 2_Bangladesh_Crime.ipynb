## b. K-means training, tuning, and visualization

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# select numeric columns
num_cols = [
    'incident_month', 'incident_week', 'weekend', 'precip',
    'visibility', 'heatindex', 'male_population', 'female_population',
    'total_population', 'gender_ration', 'average_household_size',
    'density_per_kmsq', 'literacy_rate', 'religious_institution',
    'playground', 'park', 'police_station', 'school', 'college'
]

X = df2[num_cols]

# scale data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# tune number of clusters
inertia = []
silhouette = []
k_range = range(2, 11)

for k in k_range:
    kmeans = KMeans(n_clusters=k, n_init=10)
    labels = kmeans.fit_predict(X_scaled)
    inertia.append(kmeans.inertia_)
    silhouette.append(silhouette_score(X_scaled, labels))
```

```python
# plot elbow method
plt.figure(figsize=(6,4))
plt.plot(k_range, inertia, marker='o')
plt.xlabel('number of clusters (k)')
plt.ylabel('inertia')
plt.title('elbow method')
plt.show()

# plot silhouette score
plt.figure(figsize=(6,4))
plt.plot(k_range, silhouette, marker='o')
plt.xlabel('number of clusters (k)')
plt.ylabel('silhouette score')
plt.title('silhouette analysis')
plt.show()
```

```python
# choose optimal k based on elbow & silhouette
optimal_k = 3

# train final k-means
kmeans_final = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
clusters = kmeans_final.fit_predict(X_scaled)

# attach cluster labels to dataframe
df2['cluster'] = clusters

# check cluster size
print("cluster distribution:")
print(df2['cluster'].value_counts())
```

```python
from sklearn.decomposition import PCA

# reduce dimensions to 2d for visualization
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# plot clusters
plt.figure(figsize=(7,5))
plt.scatter(X_pca[:,0], X_pca[:,1], c=df2['cluster'])
plt.xlabel('principal component 1')
plt.ylabel('principal component 2')
plt.title('k-means clustering visualization')
plt.show()
```

## c. Cluster profile

```python
profile_features = [
    'heatindex',
    'total_population',
    'density_per_kmsq',
    'literacy_rate',
    'police_station',
    'school',
    'college'
]

cluster_profile = df2.groupby('cluster')[profile_features].mean()
print(cluster_profile)
```

#### The k-means clustering results show three distinct groups based on the numerical features. Cluster 0 represents highly populated and dense areas with an average heatindex of 32.0. These areas also have relatively high literacy rates around 73.7% and substantial infrastructure, including an average of about 60 police stations, 242 schools, and 64 colleges, indicating major urban regions.

#### Cluster 1 corresponds to moderately populated regions with a higher average heatindex of 33.7, but lower population density which is around 2,674 per square kilometer. Literacy rates are significantly lower at 52.9%, and infrastructure is limited, with only approximately 4 police stations, 50 schools, and 8 colleges on average. This cluster likely represents semi-urban or peri-urban areas.

#### Cluster 2 includes less dense areas with lower average heatindex of 27.5, similar population and literacy characteristics to cluster 1, and comparable minimal infrastructure that has around 4 police stations, 49 schools, and 8 colleges. This cluster likely represents rural or low-density regions.

#### Overall, the clusters provide clear differentiation in terms of population size, density, heatindex, literacy, and local infrastructure, which can help understand regional characteristics and potentially guide further analysis or policy interventions.
